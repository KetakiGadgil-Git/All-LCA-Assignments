```python
students = {
    101:("Ketaki","CSE",85),
    102:("Aabha","CSE",78),
    103:("Neha","IT",98)
}


#add a new student record
students[104] = ("Aarnav","CSE",90)


#delete an existing student record
del students[102]


#update the details of student
students[103] = ("Neha","CSE",95)

print("Final Students Records:")

for roll_no,details in students.items():
    name,branch,marks = details
    print("Roll Number:",roll_no)
    print("Name:",name)
    print("Branch:",branch)
    print("Marks:",marks)
    print()
```
