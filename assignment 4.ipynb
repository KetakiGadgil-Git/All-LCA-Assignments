```python
import numpy as np
```

```python
rows = int(input("Enter number of rows:"))
columns = int(input("Enter number of columns:"))

print("Enter elements of first matrix:")
A = np.array([[int(input())for j in range(columns)]for i in range(rows)])

print("Enter elements of second matrix:")
B  = np.array([[int(input())for j in range(columns)]for i in range(rows)])

C = A + B

print("First Matrix:")
print(A)

print("Second Matrix:")
print(B)

print("Sum of two matrices:")
print(C)
```
